```python
#CRIAÇÃO DA TABELA BRONZE SALES TRANSACTIONS
from datetime import datetime
from pathlib import Path
import duckdb

# Procura o banco automaticamente : o notebook continuará encontrando o banco automaticamente dentro do projeto, 
# mesmo que ele seja movido para outra pasta ou máquina, desde que a estrutura do projeto seja mantida.
base_dir = Path.cwd().parent

db_path = next(base_dir.rglob("supply_chain_analytics.duckdb"))

conn = duckdb.connect(str(db_path))

# Criação da tabela bronze__sales_transactions__lite a partir da tabela raw__sales_transactions__lite
# SELECIONA AS COLUNAS DESEJADAS E CRIA A TABELA BRONZE_SALES_TRANSACTIONS
conn.execute("""
CREATE OR REPLACE TABLE bronze__sales_transactions__lite as
SELECT
    transaction_id,
    transaction_ts,
    channel,
    location_id,
    sku,
    qty,
    unit_price
FROM raw__sales_transactions__lite;
""")
# Verifica se a tabela foi criada com sucesso
tables = conn.sql("SHOW TABLES").df()

if "bronze__sales_transactions__lite" in tables["name"].values:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"✅ Tabela bronze__sales_transactions__lite criada com sucesso!"
    )
else:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"❌ Tabela bronze__sales_transactions__lite não foi criada!"
    )
conn.close()
```

```python
#CRIAÇÃO DA TABELA BRONZE LOCATION MASTER
from datetime import datetime
from pathlib import Path
import pandas as pd
import duckdb

# Procura o banco automaticamente : o notebook continuará encontrando o banco automaticamente dentro do projeto, 
# mesmo que ele seja movido para outra pasta ou máquina, desde que a estrutura do projeto seja mantida.
base_dir = Path.cwd().parent

db_path = next(base_dir.rglob("supply_chain_analytics.duckdb"))
xlsx_path = next(base_dir.rglob("raw__location_master__lite.xlsx"))
print("Loading Excel from:", xlsx_path)
df=pd.read_excel(xlsx_path)


conn = duckdb.connect(str(db_path))
conn.register("tmp__location_master__lite", df)

# Criação da tabela bronze__location_master__lite a partir da tabela raw__location_master__lite
# SELECIONA AS COLUNAS DESEJADAS E CRIA A TABELA BRONZE_LOCATION_MASTER
conn.execute("""
CREATE OR REPLACE TABLE bronze__location_master__lite as
SELECT
    location_id,
    location_name,
    location_type,
    region
FROM tmp__location_master__lite;
""")
# Verifica se a tabela foi criada com sucesso
tables = conn.sql("SHOW TABLES").df()

if "bronze__location_master__lite" in tables["name"].values:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"✅ Tabela bronze__location_master__lite criada com sucesso!"
)
else:
       print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"❌ Falha na criação da tabela bronze__location_master__lite."
    )
conn.close()
```
