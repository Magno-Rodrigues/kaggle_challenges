```python
#CRIAÇÃO DA TABELA GOLD DIM LOCATION
from datetime import datetime
from pathlib import Path
import duckdb

# Procura o banco automaticamente : o notebook continuará encontrando o banco automaticamente dentro do projeto, 
# mesmo que ele seja movido para outra pasta ou máquina, desde que a estrutura do projeto seja mantida.
base_dir = Path.cwd().parent

db_path = next(base_dir.rglob("supply_chain_analytics.duckdb"))

conn = duckdb.connect(str(db_path))

# ============================================================
# CRIAÇÃO DA DIMENSÃO DE LOCALIDADES
# ============================================================
#
# Objetivo:
# Criar a dimensão de localidades (Dim Location) a partir da
# tabela Silver de cadastro de locais.
#
# Nesta etapa os dados já passaram pelas regras de limpeza e
# padronização da camada Silver, portanto a transformação tem
# foco na modelagem analítica.
#
# Transformações realizadas:
# - Renomeação das colunas para nomenclatura de negócio
# - Estruturação da dimensão para uso em modelos analíticos
# - Disponibilização de atributos geográficos e operacionais
#
# Esta dimensão será utilizada para enriquecer tabelas fato
# de vendas, estoque e movimentações, permitindo análises por:
#
# - Região
# - Centro de distribuição
# - Loja
# - Tipo de localização
# - Unidade operacional
#
# ============================================================

conn.execute("""
CREATE OR REPLACE TABLE gold__dim_location__lite AS

SELECT

    -- Chave de negócio da localização
    location_id AS LocationID,

    -- Nome da localização
    location_name AS LocationName,

    -- Tipo da localização
    -- Ex.: Warehouse, Store, Distribution Center
    location_type AS LocationType,

    -- Região geográfica ou operacional
    region AS Region

FROM silver__location_master__lite
""")


# Verifica se a tabela foi criada com sucesso
tables = conn.sql("SHOW TABLES").df()

if "gold__dim_location__lite" in tables["name"].values:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"✅ Tabela gold__dim_location__lite criada com sucesso!"
    )
else:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"❌ Tabela gold__dim_location__lite não foi criada!"
    )
conn.close()
```

```python
#CRIAÇÃO DA TABELA GOLD DIM PRODUCT
from datetime import datetime
from pathlib import Path
import duckdb

# Procura o banco automaticamente : o notebook continuará encontrando o banco automaticamente dentro do projeto, 
# mesmo que ele seja movido para outra pasta ou máquina, desde que a estrutura do projeto seja mantida.
base_dir = Path.cwd().parent

db_path = next(base_dir.rglob("supply_chain_analytics.duckdb"))

conn = duckdb.connect(str(db_path))

# ============================================================
# CRIAÇÃO DA DIMENSÃO DE PRODUTOS
# ============================================================
#
# Objetivo:
# Criar a dimensão de produtos (Dim Product) a partir da
# tabela Silver de cadastro de produtos.
#
# Nesta etapa os dados já passaram pelos processos de limpeza,
# padronização e validação na camada Silver. O foco agora é
# disponibilizar uma estrutura otimizada para consumo analítico.
#
# Transformações realizadas:
# - Renomeação das colunas para nomenclatura de negócio
# - Organização dos atributos do produto
# - Preparação para relacionamentos com tabelas fato
#
# Esta dimensão será utilizada em análises de:
#
# - Vendas por produto
# - Vendas por categoria
# - Performance por marca
# - Giro de estoque
# - Valorização de inventário
# - Rentabilidade de produtos
#
# ============================================================

conn.execute("""
CREATE OR REPLACE TABLE gold__dim_product__lite AS

SELECT

    -- Código único do produto (SKU)
    sku AS ProductSKU,

    -- Nome do produto
    product_name AS ProductName,

    -- Categoria do produto
    category AS ProductCategory,

    -- Marca do produto
    brand AS Brand,

    -- Custo unitário utilizado em análises financeiras
    unit_cost AS UnitCost

FROM silver__product_master__lite
""")


# Verifica se a tabela foi criada com sucesso
tables = conn.sql("SHOW TABLES").df()

if "gold__dim_product__lite" in tables["name"].values:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"✅ Tabela gold__dim_product__lite criada com sucesso!"
    )
else:
    print(
        f"[{datetime.now():%d/%m/%Y %H:%M:%S}] "
        f"❌ Tabela gold__dim_product__lite não foi criada!"
    )
conn.close()
```
